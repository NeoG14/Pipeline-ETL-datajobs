## CATEGORIAS Y OFERTAS

```python
ENDPOINT = "https://www.getonbrd.com/api/v0"
```

```python
r = requests.get(f"{ENDPOINT}/categories")
```

```python
r = r.json()
```

```python
categorias = [cat["id"] for cat in r["data"]]
for i in range(len(categorias)):
        print(f"{i}: {categorias[i]}")
```

```python
for cat in r["data"]:
    print(f"ID: {cat["id"]}")
    print(f"Nombre: {cat["attributes"]["name"]}")
    print(f"Dimensions: {cat["attributes"]["dimension"]}")
    print("-"*30)
```
